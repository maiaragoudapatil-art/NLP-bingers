```python
import nltk
import re
from bs4 import BeautifulSoup

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from collections import Counter

nltk.download('punkt')
nltk.download('stopwords')
```

```python
html_data = """
<html>
<body>

<p class="tweet">I love this new phone. The battery life is amazing!</p>
<p class="tweet">This service is terrible and very disappointing.</p>
<p class="tweet">The movie was okay, not too bad but not great.</p>
<p class="tweet">Absolutely fantastic experience. Highly recommended.</p>
<p class="tweet">Worst product ever. Waste of money.</p>

</body>
</html>
"""
```

```python
soup = BeautifulSoup(html_data, "lxml")

tweets = []

for tweet in soup.find_all("p", class_="tweet"):
    tweets.append(tweet.text)

print("Extracted Tweets:")
print(tweets)
```

```python
clean_tweets = []

for tweet in tweets:
    text = re.sub(r'[^a-zA-Z\s]', '', tweet)
    clean_tweets.append(text.lower())

print("Cleaned Tweets:")
print(clean_tweets)
```

```python
import nltk
nltk.download('punkt_tab')
tokenized_tweets = []

for tweet in clean_tweets:
    words = word_tokenize(tweet)
    tokenized_tweets.append(words)

print(tokenized_tweets)
```

```python
stop_words = set(stopwords.words("english"))

filtered_tweets = []

for tweet in tokenized_tweets:
    filtered = [word for word in tweet if word not in stop_words]
    filtered_tweets.append(filtered)

print(filtered_tweets)
```

```python
positive_words = ["love", "amazing", "fantastic", "good", "great", "recommended"]
negative_words = ["terrible", "worst", "bad", "disappointing", "waste"]

for tweet in filtered_tweets:

    pos = 0
    neg = 0

    for word in tweet:
        if word in positive_words:
            pos += 1
        elif word in negative_words:
            neg += 1

    if pos > neg:
        sentiment = "Positive"
    elif neg > pos:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Tweet Words:", tweet)
    print("Sentiment:", sentiment)
    print()
```

```python
full_text = " ".join(clean_tweets)
```

```python
sentences = sent_tokenize(full_text)
```

```python
words = word_tokenize(full_text)

filtered_words = [word for word in words if word not in stop_words]

word_freq = Counter(filtered_words)

print(word_freq)
```

```python
sentences = sent_tokenize(full_text)
sentence_scores = {}

for sentence in sentences:
    for word in word_tokenize(sentence.lower()):
        if word in word_freq:
            if sentence not in sentence_scores:
                sentence_scores[sentence] = word_freq[word]
            else:
                sentence_scores[sentence] += word_freq[word]
```

```python
import heapq

summary_sentences = heapq.nlargest(2, sentence_scores, key=sentence_scores.get)

summary = " ".join(summary_sentences)

print("Summary:")
print(summary)
```
